# 05 - Model Comparison
A true Data Scientist never trusts the first model they try. Here, we will race Logistic Regression, Random Forest, and Gradient Boosting against each other.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load data
df = pd.read_csv('../data/processed/ml_features.csv', index_col=0, parse_dates=True)

feature_columns = [
    'copper_Close', 'zinc_proxy_Close', 'usdinr_Close', 'crude_Close', 'copx_Close',
    'Brass_Cost', 'Daily_Return', 'SMA_10', 'SMA_30', 'Volatility_14d'
]

X = df[feature_columns]
y = df['Target_Direction']

# Chronological split (No Time Travel!)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import joblib
import os

# Define the models we want to race
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, max_depth=3, random_state=42)
}

best_model = None
best_accuracy = 0
best_name = ""

print(" Training and Comparing Models...\n" + "-"*35)

# Loop through each model, train it, and test it
for name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)
    acc = accuracy_score(y_test, predictions)
    
    print(f" {name} Accuracy: {acc * 100:.2f}%")
    
    # Check if this is the highest score so far
    if acc > best_accuracy:
        best_accuracy = acc
        best_model = model
        best_name = name

print("-"*35)
print(f" WINNER: {best_name} with {best_accuracy * 100:.2f}% Accuracy!")

# Save ONLY the best model
os.makedirs('../models', exist_ok=True)
joblib.dump(best_model, '../models/brass_best_model.joblib')
print(f"\n Saved {best_name} to /models/brass_best_model.joblib")
```
