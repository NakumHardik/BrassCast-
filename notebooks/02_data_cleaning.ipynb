```python
import pandas as pd
import os

os.makedirs('../data/cleaned', exist_ok=True)

# 1. Load all our raw CSVs
assets = ['copper', 'zinc_proxy', 'usdinr', 'crude', 'copx']
clean_df = pd.DataFrame()

for asset in assets:
    # Read the CSV
    df = pd.read_csv(f'../data/raw/{asset}_raw.csv', header=[0, 1], index_col=0, parse_dates=True)
    
    # Grab just the simple Close price column
    close_price = df['Close'].iloc[:, 0]
    
    # Strip the timezone from the date
    close_price.index = close_price.index.tz_localize(None)
    
    # Add it to our master dataframe with a simple name
    clean_df[f"{asset}_Close"] = close_price

# 2. Forward fill the missing days (so no empty data!)
clean_df = clean_df.ffill().dropna()

# 3. Save it!
clean_df.to_csv('../data/cleaned/master_data.csv')
print("Data cleaned and saved successfully!")
clean_df.head()
```
